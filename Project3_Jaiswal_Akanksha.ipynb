```python
import pandas as pd
import glob

files = glob.glob('/content/PRSA_Data_*.csv')
print(files)

df_list = [pd.read_csv(file) for file in files]
df = pd.concat(df_list, ignore_index=True)

df.head()
```

Clean the data and check missing values for this dataset.

```python
df.dtypes
```

```python
# Check missing values column-wise
df.isnull().sum(axis=0)
```

```python
# Drop unnecessary column
df = df.drop(columns=['No'])

# Interpolate numerical columns
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
df[num_cols] = df[num_cols].interpolate(method='linear')

# Forward fill categorical column
df['wd'] = df['wd'].ffill()

# Backward fill remaining edge NaNs
df = df.bfill()

# Final check
print(df.isnull().sum())
```

```python
df.info()
df.describe()
```

Perform feature engineering on the date feature.

```python
df['datetime'] = pd.to_datetime(df[['year', 'month', 'day', 'hour']])
df = df.sort_values(by='datetime')
```

```python
# Basic time features
df['day_of_week'] = df['datetime'].dt.dayofweek   # 0=Monday
df['week_of_year'] = df['datetime'].dt.isocalendar().week
df['quarter'] = df['datetime'].dt.quarter

# Weekend indicator
df['is_weekend'] = df['day_of_week'].isin([5, 6]).astype(int)
```

```python
# Hour encoding
import numpy as np

df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)

# Month encoding
df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)

# Day encoding
df['day_sin'] = np.sin(2 * np.pi * df['day'] / 31)
df['day_cos'] = np.cos(2 * np.pi * df['day'] / 31)
```

```python
# Drop Redundant Columns
df = df.drop(columns=['year', 'month', 'day', 'hour'])
```

Split the data into 80% of training and 20% of the test dataset.

```python
df = df.sort_values(by='datetime')

split_index = int(len(df)*0.8)

train_df = df.iloc[:split_index]
test_df = df.iloc[split_index:]

X_train = train_df.drop(columns=['wd','datetime'])
y_train = train_df['wd']

X_test = test_df.drop(columns=['wd','datetime'])
y_test = test_df['wd']
```

Preprocess the data using the normalization method to convert all features into the range of [0,1]

```python
# Drop or encode categorical
X_train = X_train.drop(columns=['station'], errors='ignore')
X_test = X_test.drop(columns=['station'], errors='ignore')

# Scale
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Encode target
le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_test_enc = le.transform(y_test)
```

Build a neuron network with two hidden layers of 20 and 10 neurons to forecast wd using all other features and TensorFlow. Does it overfit or underfit the data? Please justify your answer.

```python
import tensorflow as tf
from tensorflow.keras import layers

model = tf.keras.Sequential([
    layers.Dense(20, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(10, activation='relu'),
    layers.Dense(len(le.classes_), activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(X_train_scaled, y_train_enc,
                    validation_data=(X_test_scaled, y_test_enc),
                    epochs=50, batch_size=32)
```

```python
train_loss, train_mae = model.evaluate(X_train_scaled, y_train_enc, verbose=0)
test_loss, test_mae = model.evaluate(X_test_scaled, y_test_enc, verbose=0)

print("Training Loss (MSE):", train_loss)
print("Training MAE:", train_mae)
print("Testing Loss (MSE):", test_loss)
print("Testing MAE:", test_mae)
```

**Overfitting or Underfitting? (TensorFlow)**
The model is underfitting because both training and validation accuracies are low (~18%) and very close to each other. This indicates that the model is unable to learn meaningful patterns from the data. Additionally, the loss decreases very slowly and remains high, suggesting insufficient model capacity or feature representation. Since there is no significant gap between training and validation performance, the model is not overfitting but rather underfitting.

Tune the model using the following hyperparameters using keras-tuner:
First hidden layer with units between 20 and 100 with a step size of 5
Second hidden layer with units between 10 and 50 with a step size of 2
The dropout rate for both hidden layer is between 0.2 and 0.8 with a step size of 0.1

```python
!pip install -q keras-tuner
```

```python
import keras_tuner as kt

def build_model(hp):
    model = tf.keras.Sequential()

    # First layer
    units1 = hp.Int('units1', 20, 100, step=5)
    dropout1 = hp.Float('dropout1', 0.2, 0.8, step=0.1)

    model.add(layers.Dense(units1, activation='relu'))
    model.add(layers.Dropout(dropout1))

    # Second layer
    units2 = hp.Int('units2', 10, 50, step=2)
    dropout2 = hp.Float('dropout2', 0.2, 0.8, step=0.1)

    model.add(layers.Dense(units2, activation='relu'))
    model.add(layers.Dropout(dropout2))

    model.add(layers.Dense(len(le.classes_), activation='softmax'))

    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

    return model

tuner = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=10,
    directory='tuner_dir'
)

tuner.search(X_train_scaled, y_train_enc,
             validation_data=(X_test_scaled, y_test_enc),
             epochs=20)
```

Please recommend the optimal model and specify the hyperparameters.



```python
best_hp = tuner.get_best_hyperparameters(num_trials=1)[0]

print("Best Hyperparameters:")
print("Units Layer 1:", best_hp.get('units1'))
print("Units Layer 2:", best_hp.get('units2'))
print("Dropout Layer 1:", best_hp.get('dropout1'))
print("Dropout Layer 2:", best_hp.get('dropout2'))
```

Build a neuron network with two hidden layers of 20 and 10 neurons to forecast wd using all other features and **PyTorch**. Does it overfit or underfit the data? Please justify your answer.

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Define model
class WindModel(nn.Module):
    def __init__(self, input_size, output_size):
        super(WindModel, self).__init__()

        self.model = nn.Sequential(
            nn.Linear(input_size, 20),
            nn.ReLU(),

            nn.Linear(20, 10),
            nn.ReLU(),

            nn.Linear(10, output_size)
        )

    def forward(self, x):
        return self.model(x)

# Initialize model
input_size = X_train_scaled.shape[1]
output_size = len(le.classes_)

model = WindModel(input_size, output_size)
```

```python
# Prepare data
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train_enc, dtype=torch.long)

X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test_enc, dtype=torch.long)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

train_losses = []
test_losses = []

for epoch in range(50):
    # Training
    model.train()
    optimizer.zero_grad()

    outputs = model(X_train_tensor)
    loss = criterion(outputs, y_train_tensor)

    loss.backward()
    optimizer.step()

    train_losses.append(loss.item())

    # Validation
    model.eval()
    with torch.no_grad():
        test_outputs = model(X_test_tensor)
        test_loss = criterion(test_outputs, y_test_tensor)
        test_losses.append(test_loss.item())

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Train Loss: {loss.item():.4f}, Test Loss: {test_loss.item():.4f}")
```

```python
# Evaluate accuracy
from sklearn.metrics import accuracy_score

model.eval()
with torch.no_grad():
    preds = model(X_test_tensor)
    predicted = torch.argmax(preds, axis=1)

accuracy = accuracy_score(y_test_tensor, predicted)
print("Test Accuracy:", accuracy)
```

The model is underfitting because both training and test losses are high and similar, and the accuracy is very low (~14%), indicating that the model fails to learn meaningful patterns from the data.
